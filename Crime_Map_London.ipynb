# **Uploading files & combining data tables**

```python
# Importing tools
import pandas as pd
import glob
import numpy as np
from google.colab import files

# Upload Crime Data
# Source: https://data.police.uk/data/
uploaded = files.upload()
```

```python
# Checking files
csv_files = list(uploaded.keys())
print(csv_files)
```

```python
# Combining all data into one file
df_list = [pd.read_csv(f) for f in csv_files]
combined_df = pd.concat(df_list, ignore_index=True)
combined_df.head()
```

# Removing Duplicates (Rows with NaN Crime Data are not included)

```python
# Check total number of rows
print('Total Rows:', combined_df.shape)
```

```python
# Check how many rows are duplicates based on Crime ID
print(combined_df['Crime ID'].duplicated().sum())
```

```python
# Separate the data into two tables - thoses with a Crime ID and those without one. Rechecking duplicates for data with a Crime ID
has_id = combined_df[combined_df['Crime ID'].notna()].copy()
no_id = combined_df[combined_df['Crime ID'].isna()].copy()
print(has_id['Crime ID'].duplicated().sum())

# Checking table size again
print('Rows with Crime ID:', has_id.shape)
print('Rows with NaN Crime ID:', no_id.shape)
```

```python
# Removing duplicates from the has_id table
combined_df['Month'] = pd.to_datetime(combined_df['Month'], format='%Y-%m')
has_id = has_id.drop_duplicates(subset='Crime ID', keep='last')
print('After deduplicate:', has_id.shape)
```

```python
# Combining tables
combined_df = pd.concat([has_id, no_id], ignore_index=True)
print('Final:', combined_df.shape)
```

```python
# Checking the table
combined_df.head()
```

# Filtering data based on a 0.5 KM radius from London Office (Using haversine)



```python
def haversine_dis(lat1, long1, lat2, long2):
  # Covert degrees to radians
  lat1, long1, lat2, long2 = map(np.radians, [lat1, long1, lat2, long2])

  # Differences
  dlat = lat2 - lat1
  dlong = long2 - long1

  # Haversine formuala
  a = np.sin(dlat / 2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlong / 2)**2
  c = 2 * np.arcsin(np.sqrt(a))

  r = 6371 # Radius of Earth in KM
  return r * c

# Lat/Long of London Office
Lon_lat = 51.5068
Lon_long = -0.1310

# Calulating the distance of every crime data point to the London Office. The calculated KM stored into a new column
combined_df['distance_km'] = haversine_dis(
      Lon_lat, Lon_long, combined_df['Latitude'], combined_df['Longitude']
).round(2)

print(combined_df.shape)
combined_df.head()
```

```python
# Filter all crime outside of a 0.5 KM radius
combined_df = combined_df[combined_df['distance_km'] <= 0.5]

# Check table again
print(combined_df.shape)
combined_df.head()
```

# Export table as a CSV

```python
out_path = "London_Office_Crime_Data_2025_-_2026.csv"
combined_df.to_csv(out_path, index=False)
```
